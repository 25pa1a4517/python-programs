```python
#tuple operations
t1=(28,"nivas")
t2=(24,"nivas2")
print(t1+t2) #concatenation
print(t1*2) #repetation
print(28 in t1)#membership
print(len(t1))#length
```

```python
#tuple methods
t1=(1,2,3,4,5)
print(t1.count(3))#count
print(t1.index(3))#index
```

```python
#zip func
t1=(1,2,3)
t2=("nivas","nivas2","nivas3")
print(tuple(zip(t1,t2)))
```

SETS IN python


```python
S=set()#creates empty set
s={1,2,3,11,11,34,24,28}
n={3,5,15,16,56}
print(s|n)
print(s&n)
print(s-n)
print(s^n)
```

```python
#methods in sets
s={1,3,24,28}
n={3,5,15,16,56}
print(s.union(n))
print(s.intersection(n))
print(s.difference(n))
print(s.symmetric_difference(n))
s.add(56)
print(s)
s.update([10,20,30])
print(s)
s.remove(3)
print(s)
s.discard(24)
print(s)
s.pop()
print(s)
s.clear()
print(s)
```
