```python
#program that returns multiple values
def multiple_function(*num):
  s=sum(num)
  c=len(num)
  avg=s/c
  return s,c,avg
res_sum,res_count,res_avg=multiple_function(1,2,3,4,5)
print("the sum is",res_sum,"the count is",res_count,"the average is",res_avg)
```

```python
#program that prints reverse of the number using functions
def reverse(num):
  rev=0
  while (num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  return rev
print(reverse(12345))
```

```python
#program that count the no of vowels and consonants in a string
```

```python
a=str(input())
b=a.lower()
vowels="aieou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("no of vowels",count_v)
print("no of consonants",count_c)
```

```python
#program that checks palindrome or not using functions
def palindrome(num):
  original=num
  rev=0
  while(num!=0):
    rem=num%10
    rev=rev*10+rem
    num=num//10
  if(rev==original):
    return True
  else:
    return False
print(palindrome(545))
print(palindrome(23345))
```
