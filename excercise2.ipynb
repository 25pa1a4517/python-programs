```python
#2.leap year
year=int(input())
x=year
if((x%400==0) or ((x%4==0) and (x%100!=0))):
  print(f"{year} is a leap year")
else:
  print(f"{year} is not leap year")
```

```python
#3.match case
day=int(input())
match day:
  case 1:
    print("sunday")
  case 2:
    print("monday")
  case 3:
    print("tuesday")
  case 4:
    print("wednesday")
  case 5:
    print("thursday")
  case 6:
    print("friday")
  case 7:
    print("saturday")
  case _:
    print("invalid")
```

```python
#sum of the digits
num=int(input())
sum=0
while(num>0):
   rem=num%10
   sum+=rem
   num//=10
print("sum of digits:",sum)
```

```python
#reverse of a num
num=int(input())
sum=0
while(num>0):
   rem=num%10
   sum=sum*10+rem
   num//=10
print(sum)
```

```python
#prime or not
num = int(input())
if num <= 1:
    flag = 1
elif num == 2:
    flag = 0
else:
    flag = 0
    i = 2
    while i < num:
        if num % i == 0:
            flag = 1
            break
        i += 1

print("prime" if flag == 0 else "not prime")
```
