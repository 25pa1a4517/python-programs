```python
#write a program to sort words and if any upper case change to lower case and copy to another file
with open("new.txt","r") as f:
  data=f.read().split()
  data_lower=[i.lower() for i in data]
  data_lower.sort()
  with open("new2.txt","w") as f2:
    for i in data_lower:
      f2.write(i+"\n")
```

```python
#write a program that computes the number of characters,words and lines in a file
with open("new.txt","r") as f:
  lines=f.readlines()
  print(len(lines))
  #number of characters
  chars=0
  for i in lines:
    chars+=len(i)
  print(chars)
  #number of words
  words=0
  for i in lines:
    words+=len(i.split())
  print(words)
```

```python
#Write a program to sort words in a file and put them in another file. The output
#file should have only lower-case words, so any upper-case words from source
#must be lowered.
```

```python

# Create source file
with open("f1.txt", "w") as f:
    f.write("Banana Apple Orange Mango")

# Read the source file
with open("f1.txt", "r") as f:
    data = f.read().split()

# Convert all words into lowercase
data_lower = [i.lower() for i in data]

# Sort the words
sort_data = sorted(data_lower)

# Write sorted words into another file
with open("copy.txt", "w") as f2:
    for i in sort_data:
        f2.write(i + "\n")

print("Done")
```

```python
#program on basic use of rb,wb,ab binary files
with open("download.jpg","rb") as f:
  data=f.read()
with open("download2.jpg","wb") as f2:
    f2.write(data)
print("new binary file created and pic copied")
```

```python
#pickle package
import pickle
with open("s1.dat","wb") as f:
  data=[1,2,3,"hello"]
  pickle.dump(data,f)
with open("s1.dat","rb") as f2:
  d=pickle.load(f2)
  print(d)
```
